## Univariate Analysis

```python
# Histogram for active users

sns.histplot(x=df["active_users"])
plt.title("Distribution of Active Users")
plt.xlabel("Active Users")
plt.ylabel("Frequency")
plt.show()
```

```python
# Histogram for CPU usage

sns.histplot(x=df["cpu_usage_percent"], kde=True)
plt.title("Distribution of CPU Usage")
plt.xlabel("CPU Usage (%)")
plt.ylabel("Frequency")
plt.show()
```

```python
# Histogram for memory usage

sns.histplot(x=df["memory_usage_percent"], kde=True)
plt.title("Distribution of Memory Usage")
plt.xlabel("Memory Usage (%)")
plt.ylabel("Frequency")
plt.show()
```

```python
# Boxplot for response time

sns.boxplot(x=df["avg_response_time_ms"])
plt.title("Distribution of Average Response Time")
plt.xlabel("Average Response Time (ms)")
plt.show()
```

```python
# Boxplot for error rate

sns.boxplot(x=df["error_rate_percent"])
plt.title("Distribution of Error Rate")
plt.xlabel("Error Rate (%)")
plt.show()
```

```python
# Countplot for weekend and non-weekend records

sns.countplot(x=df["is_weekend"])
plt.title("Weekend vs Non-Weekend")
plt.xlabel("Is Weekend")
plt.ylabel("Count")
plt.show()
```

```python
# Countplot for peak hour and non-peak hour records

sns.countplot(x=df["is_peak_hour"])
plt.title("Peak Hour vs Non-Peak Hour")
plt.xlabel("Is Peak Hour")
plt.ylabel("Count")
plt.show()
```

## Bivariate Analysis
#### Numerical vs Numerical

```python
# Scatterplot between active users and CPU usage

sns.scatterplot(x=df["active_users"], y=df["cpu_usage_percent"])
plt.title("Active Users vs CPU Usage")
plt.xlabel("Active Users")
plt.ylabel("CPU Usage (%)")
plt.show()
```

```python
# Scatterplot between active users and memory usage

sns.scatterplot(x=df["active_users"], y=df["memory_usage_percent"])
plt.title("Active Users vs Memory Usage")
plt.xlabel("Active Users")
plt.ylabel("Memory Usage (%)")
plt.show()
```

```python
# Scatterplot between CPU usage and response time

sns.scatterplot(x=df["cpu_usage_percent"], y=df["avg_response_time_ms"])
plt.title("CPU Usage vs Response Time")
plt.xlabel("CPU Usage (%)")
plt.ylabel("Average Response Time (ms)")
plt.show()
```

### Categorical vs Numerical

```python
# Boxplot for CPU usage during weekend and non-weekend

sns.boxplot(x=df["is_weekend"], y=df["cpu_usage_percent"])
plt.title("Weekend vs CPU Usage")
plt.xlabel("Is Weekend")
plt.ylabel("CPU Usage (%)")
plt.show()
```

```python
# Barplot for average CPU usage during weekend and non-weekend

sns.barplot(x=df["is_weekend"], y=df["cpu_usage_percent"])
plt.title("Weekend vs Average CPU Usage")
plt.xlabel("Is Weekend")
plt.ylabel("Average CPU Usage (%)")
plt.show()
```

```python
# Boxplot for CPU usage during peak and non-peak hours

sns.boxplot(x=df["is_peak_hour"], y=df["cpu_usage_percent"])
plt.title("Peak Hour vs CPU Usage")
plt.xlabel("Is Peak Hour")
plt.ylabel("CPU Usage (%)")
plt.show()
```

### Multivariate Analysis

```python
# Active users vs CPU usage according to peak hour

sns.scatterplot(x=df["active_users"],y=df["cpu_usage_percent"],hue=df["is_peak_hour"])

plt.title("Active Users vs CPU Usage as per Peak Hour")
plt.xlabel("Active Users")
plt.ylabel("CPU Usage (%)")
plt.show()
```

```python
# Response time vs error rate according to peak hour

sns.scatterplot(x=df["avg_response_time_ms"],y=df["error_rate_percent"],hue=df["is_peak_hour"])

plt.title("Response Time vs Error Rate as per Peak Hour")
plt.xlabel("Average Response Time (ms)")
plt.ylabel("Error Rate (%)")
plt.show()
```

```python
# Requests per minute vs CPU usage according to peak hour

sns.scatterplot(x=df["requests_per_minute"],y=df["cpu_usage_percent"],hue=df["is_peak_hour"])

plt.title("Requests per Minute vs CPU Usage as per Peak Hour")
plt.xlabel("Requests per Minute")
plt.ylabel("CPU Usage (%)")
plt.show()
```
