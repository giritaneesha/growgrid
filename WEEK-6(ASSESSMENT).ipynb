# ASSESSMENT QUESTIONS

```python
# 1. Which diagnoses account for the highest number of unique prescriptions, and how large is the gap between the leading diagnoses and the rest?

diagnosis_count = df.groupby("Diagnosis")["Prescription_ID"].nunique().sort_values(ascending=False)

print("Top 5 diagnoses by number of prescriptions:")
print(diagnosis_count.head(5))

print("\nTop diagnosis:", diagnosis_count.index[0])
print("Number of prescriptions:", diagnosis_count.iloc[0])

gap = diagnosis_count.iloc[0] - diagnosis_count.iloc[1]
print("Gap with second-highest diagnosis:", gap)
```

```python
diagnosis_count.head(10).plot(kind="bar")
plt.title("Top Diagnoses by Prescription Count")
plt.xlabel("Diagnosis")
plt.ylabel("Number of Prescriptions")
plt.xticks(rotation=90)
plt.show()
```

```python
# 2. Which departments handle the highest prescription volume, and what patterns distinguish high-volume departments from low-volume departments?

department_count = df.groupby("Department")["Prescription_ID"].nunique().sort_values(ascending=False)

print("Prescription count by department:")
print(department_count)

print("\nHighest-volume department:", department_count.index[0])
print("Prescriptions:", department_count.iloc[0])

print("\nLowest-volume department:", department_count.index[-1])
print("Prescriptions:", department_count.iloc[-1])
```

```python
department_count.plot(kind="bar")
plt.title("Prescription Volume by Department")
plt.xlabel("Department")
plt.ylabel("Number of Prescriptions")
plt.xticks(rotation=90)
plt.show()
```

```python
# 3. Which departments generate the highest total medicine expenditure, and does this ranking match their prescription volume?

department_cost = df.groupby("Department")["Total_Medicine_Cost"].sum().sort_values(ascending=False)

print("Total medicine expenditure by department:")
print(department_cost)

print("\nDepartment with highest expenditure:", department_cost.index[0])
print("Total expenditure:", round(department_cost.iloc[0], 2))
```

```python
department_cost.plot(kind="bar")
plt.title("Medicine Expenditure by Department")
plt.xlabel("Department")
plt.ylabel("Total Medicine Expenditure")
plt.xticks(rotation=90)
plt.show()
```

```python
# 4. Which departments have the highest average cost per prescription, and what may be driving these differences?

department_cost = df.groupby("Department").agg(
    Total_Cost=("Total_Medicine_Cost", "sum"),
    Prescriptions=("Prescription_ID", "nunique")
)

department_cost["Average_Cost"] = department_cost["Total_Cost"] / department_cost["Prescriptions"]

department_cost = department_cost.sort_values("Average_Cost", ascending=False)

print("Average prescription cost by department:")
print(department_cost)

print("\nHighest average cost department:", department_cost.index[0])
print("Average prescription cost:", round(department_cost.iloc[0]["Average_Cost"], 2))
```

```python
department_cost["Average_Cost"].plot(kind="bar")
plt.title("Average Prescription Cost by Department")
plt.xlabel("Department")
plt.ylabel("Average Prescription Cost")
plt.xticks(rotation=90)
plt.show()
```

```python
# 5. How does the average number of medicines included in a prescription vary across departments?

medicine_count = df.groupby(["Department", "Prescription_ID"])["Medicine_Name"].nunique()
average_medicine = medicine_count.groupby("Department").mean().sort_values(ascending=False)

print("Average number of medicines per prescription:")
print(average_medicine)

print("\nHighest average:", average_medicine.index[0])
print("Average medicines:", round(average_medicine.iloc[0], 2))
```

```python
average_medicine.plot(kind="bar")
plt.title("Average Medicines per Prescription by Department")
plt.xlabel("Department")
plt.ylabel("Average Number of Medicines")
plt.xticks(rotation=90)
plt.show()
```

```python
# 6. Which departments prescribe the widest variety of medicine categories?

category_count = df.groupby("Department")["Medicine_Category"].nunique().sort_values(ascending=False)

print("Number of medicine categories by department:")
print(category_count)

print("\nDepartment with widest variety:", category_count.index[0])
print("Number of categories:", category_count.iloc[0])
```

```python
category_count.plot(kind="bar")
plt.title("Medicine Category Variety by Department")
plt.xlabel("Department")
plt.ylabel("Number of Medicine Categories")
plt.xticks(rotation=90)
plt.show()
```

```python
# 7. What is the most common diagnosis within each department, and are any diagnoses concentrated in specific departments?

department_diagnosis = df.groupby(["Department", "Diagnosis"])["Prescription_ID"].nunique().reset_index()

common_diagnosis = department_diagnosis.loc[
    department_diagnosis.groupby("Department")["Prescription_ID"].idxmax()
]

print("Most common diagnosis in each department:")
print(common_diagnosis)
```

```python
plt.figure(figsize=(12,6))
sns.countplot(x=df["Department"])
plt.title("Prescription Distribution Across Departments")
plt.xlabel("Department")
plt.ylabel("Count")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# 8. Which diagnosis generates the highest medicine expenditure within each department?

department_diagnosis_cost = df.groupby(["Department", "Diagnosis"])["Total_Medicine_Cost"].sum().reset_index()

highest_cost = department_diagnosis_cost.loc[
    department_diagnosis_cost.groupby("Department")["Total_Medicine_Cost"].idxmax()
]

print("Highest expenditure diagnosis in each department:")
print(highest_cost)
```

```python
# 29. Which medicines have high unit prices but relatively low prescription frequency?

medicine_price = df.groupby("Medicine_Name").agg(
    Unit_Price=("Unit_Price", "mean"),
    Prescriptions=("Prescription_ID", "nunique")
)

high_price_low_use = medicine_price[
    (medicine_price["Unit_Price"] > medicine_price["Unit_Price"].median()) &
    (medicine_price["Prescriptions"] < medicine_price["Prescriptions"].median())
].sort_values("Unit_Price", ascending=False)

print("High-price and low-frequency medicines:")
print(high_price_low_use.head(10))
```

```python
high_price_low_use.head(10)["Unit_Price"].plot(kind="bar")
plt.title("High Price and Low Frequency Medicines")
plt.xlabel("Medicine")
plt.ylabel("Average Unit Price")
plt.xticks(rotation=90)
plt.show()
```

```python
# 30. Which medicines have low unit prices but generate high expenditure because of large quantities or frequent use?

medicine_use = df.groupby("Medicine_Name").agg(
    Unit_Price=("Unit_Price", "mean"),
    Quantity=("Quantity", "sum"),
    Prescriptions=("Prescription_ID", "nunique"),
    Expenditure=("Total_Medicine_Cost", "sum")
)

low_price_high_cost = medicine_use[
    (medicine_use["Unit_Price"] < medicine_use["Unit_Price"].median()) &
    (medicine_use["Expenditure"] > medicine_use["Expenditure"].median())
].sort_values("Expenditure", ascending=False)

print("Low-price medicines with high expenditure:")
print(low_price_high_cost.head(10))
```

```python
# Create a prescription and medicine basket
basket = df.groupby(["Prescription_ID", "Medicine_Name"])["Medicine_Name"].count().unstack(fill_value=0)

# Display the first five rows
basket.head()
```

## Convert Basket into Boolean Values

```python
# Convert medicine counts into True and False
basket = basket.astype(bool)

# Display the Boolean basket
basket.head()
```

```python
# Check the number of transactions and medicines
basket.shape
```

# Install and Import Apriori Tools

```python
# Install the mlxtend library
%pip install -q mlxtend
```

```python
# Import the classic Apriori function for frequent-itemset discovery.
from mlxtend.frequent_patterns import apriori

# Import the analytical association-rules function for rule generation.
from mlxtend.frequent_patterns import association_rules
```

# Apply Apriori Algorithm

```python
# Find frequent itemsets using 2% minimum support
frequent_itemsets = apriori(basket, min_support=0.02, use_colnames=True)

# Display the frequent itemsets
frequent_itemsets.head()
```

### Observation

Apriori has identified medicines and medicine combinations that occur frequently enough to meet the minimum support of 0.02.

## Frequent Itemsets by Support

```python
# Sort frequent itemsets from highest to lowest support
frequent_itemsets = frequent_itemsets.sort_values("support", ascending=False)

# Display the top 20 frequent itemsets
frequent_itemsets.head(20)
```

# Generate Association Rules

The frequent itemsets will now be converted into association rules.A rule can be written as:

**Medicine A → Medicine B**

This means that Medicine B is found when Medicine A is present.

```python
# Generate association rules using lift
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.0)

# Display the first few rules
rules.head()
```

## Association Rules Sorted by Lift

```python
# Sort association rules from highest to lowest lift
rules = rules.sort_values("lift", ascending=False)

# Display the top 15 rules
rules.head(15)
```

# Support, Confidence and Lift

```python
# filter rules that show both meaningful confidence and strong positive lift.
strong_rules = rules[(rules['confidence'] >= 0.50) & (rules['lift'] > 1.20)].copy()

# sort the filtered rules from highest to lowest lift.
strong_rules = strong_rules.sort_values(by=['lift', 'confidence'], ascending=False)

# display the key metrics for the strongest filtered rules.
strong_rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(30)
```

### Convert Association Rules into Readable Format

```python
# Convert antecedents into readable text
strong_rules["antecedents"] = strong_rules["antecedents"].apply(lambda x: ", ".join(sorted(x)))

# Convert consequents into readable text
strong_rules["consequents"] = strong_rules["consequents"].apply(lambda x: ", ".join(sorted(x)))

# Create an easy-to-read rule
strong_rules["Rule"] = strong_rules["antecedents"] + " → " + strong_rules["consequents"]

# Display the readable rules
strong_rules[["Rule", "support", "confidence", "lift"]].head(15)
```

# Visualisation of Association Rules

```python
# Select the top 15 rules
top_rules = strong_rules.head(15)

# Plot the rules by lift
plt.figure(figsize=(12, 8))
sns.barplot(data=top_rules, x="lift", y="Rule")
plt.title("Top Apriori Association Rules by Lift")
plt.xlabel("Lift")
plt.ylabel("Association Rule")
plt.tight_layout()
plt.show()
```

## Support vs Confidence

```python
# Plot support against confidence
plt.figure(figsize=(10, 6))
sns.scatterplot(data=rules, x="support", y="confidence", size="lift", sizes=(30, 300), alpha=0.7)
plt.title("Apriori Rules: Support vs Confidence")
plt.xlabel("Support")
plt.ylabel("Confidence")
plt.tight_layout()
plt.show()
```
