```python
import pandas as pd
df = pd.read_csv('/Users/keva/Downloads/finance_loan_portfolio_python_dataset.csv')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#1 Age
sns.histplot(x=df["Age"], kde=True)
plt.title("Histogram for Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
#2 Annual Income
sns.histplot(x=df["Annual_Income_INR"], kde=True)
plt.title("Histogram for Annual Income")
plt.xlabel("Annual Income")
plt.ylabel("Frequency")
plt.show()
```

```python
#3 Credit Score
sns.histplot(x=df["Credit_Score"], kde=True)
plt.title("Histogram for Credit Score")
plt.xlabel("Credit Score")
plt.ylabel("Frequency")
plt.show()
```

```python
#4 Gender
sns.countplot(x = df["Gender"])
plt.title("Count of Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
#5 Loan Type 
sns.countplot(x = df["Loan_Type"])
plt.title("Count of Loan Type ")
plt.xlabel("Loan Type ")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()
```

Bivariate_Analysis

```python
#1.Bar Plot(Categorial vs Numerical)
#Gender vs Annual Income

sns.boxplot(x=df["Gender"], y=df["Annual_Income_INR"])
plt.title("Gender wise Annual Income")
plt.xlabel("Gender")
plt.ylabel("Annual Income")
plt.show()
```

```python
#2. Box Plot (Categorical vs Numerical)
#Loan Type vs Loan Amount

sns.boxplot(x=df["Loan_Type"], y=df["Loan_Amount_INR"])
plt.title("Loan Type vs Loan Amount")
plt.xlabel("Loan Type")
plt.ylabel("Loan Amount")
plt.xticks(rotation=90)
plt.show()
```

```python
#3. Bar Plot(Categorical vs Numerical)
#Risk Category vs Credit Score

sns.barplot (x=df ["Risk_Category"], y=df["Credit_Score"])
plt.title("Risk Category vs Credit Score")
plt.xlabel("Risk Category")
plt.ylabel ("Credit Score")
plt.show()
```

```python
#Scatter plot (Numerical vs Numerical)
#Annual Income vs Loan Amount
sns.scatterplot(x=df["Annual_Income_INR"], y=df ["Loan_Amount_INR"])
plt.title("Annual Income vs Loan Amount")
plt.xlabel("Annual Income")
plt.ylabel("Loan Amount")
plt.show()
```

```python
#5. Scatter Plot (Numerical vs Numerical)
#Age vs Credit Score
sns.scatterplot(x=df["Age"], y=df["Credit_Score"])
plt.title("Age vs Credit Score") 
plt.xlabel("Age")
plt.ylabel("Credit Score")
```

```python
#6. Scatter Plot (Numerical vs Numerical)
#Loan Amount vs EMI

sns.scatterplot(x=df["Loan_Amount_INR"], y=df ["Monthly_EMI_INR"])
plt.title("Loan Amount vs Monthly EMI INR")
plt.xlabel("Loan Amount")
plt.ylabel("Monthly_EMI_INR")
plt.show()
```

```python
#Multivariate Analysis
```

```python
#1. Scatter Plot (Numerical vs Numerical + hue)
# Annual Income vs Loan Amount (Loan Type)

sns. scatterplot(x=df ["Annual_Income_INR"], y=df ["Loan_Amount_INR"], hue=df ["Loan_Type"])
plt.title("Annual Income vs Loan Amount by Loan Type") 
plt.xlabel("Annual Income")
plt.ylabel("Loan Amount")
plt.show()
```

```python
# 2. Scatter Plot (Numerical vs Numerical + hue)
# Age vs Credit Score (Risk Category)

sns.scatterplot(x=df ["Age"], y=df ["Credit_Score"], hue=df ["Risk_Category"])
plt.title("Age vs Credit Score by Risk Category") 
plt.xlabel("Age")
plt.ylabel("Credit Score")
plt.show()
```

```python
# 3. Bar Plot (Categorical vs Numerical)
# Loan Type vs Annual Income (Risk Category)

sns. barplot (x=df ["Loan_Type"] ,y=df ["Annual_Income_INR"] , hue=df ["Risk_Category"],errorbar=None)
plt. title("Loan Type vs Annual Income by Risk Category")
plt.xlabel("Loan Type")
plt.ylabel( "Annual Income")
plt.xticks(rotation=90)
plt.show()
```

```python
#4. Count Plot (Categorical vs Categorical)
# City vs Loan Type
sns. countplot(x=df ["City"], hue=df ["Loan_Type" ])
plt. title("City vs Loan Type")
plt.xlabel ("City")
plt.ylabel( "Count" )
plt.xticks(rotation=90)
plt.show()
```

```python
#5. Box Plot (Categorical vs Numerical)
# Loan Type vs Loan Amount (Gender)

sns. boxplot (x=df ["Loan_Type"], y=df ["Loan_Amount_INR"], hue=df ["Gender"]) 
plt. title("Loan Type vs Loan Amount by Gender")
plt.xlabel("Loan Type") 
plt. ylabel("Loan Amount") 
plt.xticks(rotation=90) 
plt.show()
```
