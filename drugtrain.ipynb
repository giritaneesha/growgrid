## A Data Analysis and Multi-Class Classification Project

# 1. Project Introduction

Drug prescription plays an important role in healthcare, as the appropriate medicine may vary depending on the characteristics of a patient.

This project focuses on analyzing patient information and identifying patterns in drug prescriptions. The dataset contains patient characteristics such as age, sex, blood pressure, cholesterol level, and the Sodium-to-Potassium (Na_to_K) ratio.

The main objective of this project is to understand the prescription patterns through Exploratory Data Analysis (EDA) and build a machine learning classification model that can predict the appropriate drug category for a patient.

The project follows a complete data science workflow, starting from data acquisition and data understanding, followed by data preprocessing, exploratory data analysis, visualization, feature preparation, model building, model evaluation, and final prediction.

# 2. Problem Statement

The objective of this project is to analyze patient-related information and predict the drug prescribed to a patient based on their characteristics.

The dataset contains different patient attributes such as:

- Age
- Sex
- Blood Pressure (BP)
- Cholesterol
- Na_to_K ratio

The target variable is the **Drug** category.

This is a **Multi-Class Classification** problem because the target variable contains multiple drug categories.

The project will first analyze the relationship between patient characteristics and drug prescriptions using Exploratory Data Analysis (EDA). After that, machine learning classification models will be trained and evaluated to identify the model that performs best in predicting the prescribed drug.
