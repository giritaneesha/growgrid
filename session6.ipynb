```python
#1.remove all consonants from each string

words=["table","chair","glass"]
vowels="aeiouAEIOU"

for word in words:
    result = " "

    for ch in word:
        if ch in vowels:
            result = result + ch
            print(result)
    
```

```python
#3.string containing atleast 3 different vowels

words=["environment","grass","peacock","laptop"]
vowels="aeiou"

for word in words:
    count=0
    for v in vowels:
        if v in word:
            count=count+1
    
    if count>=3: 
        print(word)
```

```python
#8.palindrome

words = ["malayalam","level","madam","class"]

for i in range(len(words)):
    if words[i] == words[i][::-1]:
        print((i,words[i]))
```

```python
#10.filter string that match their sorted version

words = ["abc","bca","hijk","bac","rst"]

for word in words:
    if word == "".join(sorted(word)):
        print(word)
```

```python
#2. Filter strings that alternate vowels and consonants

words = ["hello","python","banana","code"]

def alternate(x):
    for i in range(len(x)-1):
        if (x[i].lower() in "aeiouAEIOU")== (x[i+1].lower() in "aeiouAEIOU"):
            return False
    return True

result = list(filter(alternate,words))
print(result)
```

```python
#4. Built a string of the first character from each word

words = ["apple","banana","mango","orange"]
def first_character(x):
    return x[0]

result = "".join(map(first_character, words))
    
print(result)
```

```python
#5. Reduce list to a string of only digits extracted from all words

words = ["hi03","red19","gt65"]
result = ""

for x in words:
    for ch in x:
        if ch.isdigit():
            result += ch
            print(result)
```

```python
#6.double the word at every index divisible by 4
words = ["apple", "bag", "cat", "dog", "eat", "feet", "great"]

def change(x):
    if x[0] % 4 == 0:
        return x[1] * 2
    else:
        return x[1]

result = list(map(change, enumerate(words)))
print(result)
```

```python
#7.return all words that appear more than once 
sentence = "this is a table this is spoon"
words = sentence.split()

def repeated(x):
    return words.count(x) > 1

result = list(filter(repeated, set(words)))
print(result)
```

```python
#9.sort by proximity to target length(5)
words = ["hi", "apple", "banana", "cat", "python"]

def length_difference(x):
    return (len(x) - 5) ** 2

result = sorted(words,key=length_difference)
print(result)
```

```python
#11.Capitalize all words

words = ["apple", "banana", "cat"]

def cap(x):
    return x.capitalize()
result = list(map(cap, words))

print(result)
```

```python
#12.Convert strings to their lengths

words = ["apple", "banana", "cat"]
result = list(map(len, words))
print(result)
```

```python
#13.Filter palindrome strings

words = ["malayalam","level","madam","earth"]
def palindrome(x):
    return x == x[::-1]

result = list(filter(palindrome, words))

print(result)
```

```python
#14.Concatenate all strings into one

words = ["My","name","is","taneesha"]
result = "".join(words)
print(result)
```

```python
#15.Find the longest string

words = ["jupiter","mars","earth","moon","solarsystem"]
longest = words[0]

for x in words:
    if len(x) > len(longest):
        longest = x

print(longest)
```

```python
#17.Find index of the longest string
words = ["apple", "banana", "cat", "dog","egg"]

index = 0
for i in range(len(words)):
    if len(words[i]) > len(words[index]):
        index = i

print(index)
```

```python
#18.Reverse each string
words = ["apple", "banana", "cat"]
result = list(map(lambda x: x[::-1], words))
print(result)
```

```python
#19.Get ASCII values of characters in a string

words = "ABC"
result = list(map(ord, words))
print(result)
```

```python
#20.Filter strings that start with a vowel

words = ["banana", "orange", "cat", "evnironment"]

result = list(filter(lambda x: x[0].lower() in "aeiou", words))

print(result)
```

```python
#16.Create list of (index, string) for strings starting with ‘b’

words = ["ball","apple", "bat","basket","boy","baby", "cat"]

def check(x):
    return x.startswith("b")

result = list(filter(check, words))
print(result)
```
